```python
import pandas as pd
import numpy as np
```

```python
df = pd.DataFrame(
  [
   [20, 40, 520], [5, 10, 125], [8, 60, 380],
    [22, 35, 510], [10, 55, 370], [3, 15, 105],
    [12, 45, 340], [18, 50, 470], [15, 5, 195],
    [4, 30, 180], [6, 0, 150], [17, 40, 450],
    [3, 35, 185], [14, 30, 350], [6, 20, 190],
    [10, 50, 320], [14, 10, 210], [2, 12, 110],
    [20, 45, 530], [1, 35, 155],
  ],
  columns=['grasa', 'carbohidrato', 'calorias']

)
print(df)
```

```python
#definimos las variables que serán tomadas del dataframe
valores_grasa = df['grasa']
valores_carbohidrato  = df['carbohidrato']
valores_calorias = df['calorias']


#iniciamos pesos
#es un numero que la neurona ajusta para aprender
w1 = 0
w2 = 0

#iniciamos bias
#es un numero que le da flexibilidad a la neurona , es como el intercepto de la line a recta y=mx+b
b = 0

#calculo de los valores
def salida_calculada(grasa, carbos, w1, w2, b):
    return grasa * w1 + carbos * w2 + b
#calculo de todas la variables
def salidas_todas(v_grasa, v_carbos, w1, w2, b):
    calculados = []
    for i in range(len(v_grasa)):
        calculado = salida_calculada(v_grasa[i], v_carbos[i], w1, w2, b)
        calculados.append(calculado)
    return calculados

#Error, que tan mal lo esta haciendo la neurona

def funcion_costos(reales, calculados):
    suma = 0
    for i in range(len(reales)):
        suma = suma + (reales[i] - calculados[i])**2
    N = len(reales)
    return (1 / (2 * N)) * suma
#La derivada le dice en que direccion mover cada peso para equivocarse menos

def dW1(reales, calculados, grasa):
    suma = 0
    for i in range(len(calculados)):
        error_i = reales[i] - calculados[i]   # ¿cuánto me equivoqué en esta fila?
        suma = suma + error_i * (-1 * grasa[i]) # ¿cuánto culpo a w1 de ese error?
    return (1 / len(reales)) * suma    #promedio

def dW2(reales, calculados, carbos):
    suma = 0
    for i in range(len(calculados)):
        suma = suma + (reales[i] - calculados[i]) * (-1 * carbos[i])
    return (1 / len(reales)) * suma
def dB(reales, calculados):
    suma = 0
    for i in range(len(calculados)):
        suma = suma + (reales[i] - calculados[i]) * (-1)
    return (1 / len(reales)) * suma



def entrenar(v_grasa, v_carbos, reales, epocas=50000, lr=0.001):
    w1, w2, b = 0, 0, 0
    for epoca in range(epocas):
        calculados = salidas_todas(v_grasa, v_carbos, w1, w2, b)
        w1 = w1 - lr * dW1(reales, calculados, v_grasa)
        w2 = w2 - lr * dW2(reales, calculados, v_carbos)
        b  = b  - lr * dB(reales, calculados)
        if epoca % 100 == 0:
            error = funcion_costos(reales, calculados)
            print(f"Época {epoca:4d} → Error: {error:,.2f}")
    print(f"\nPesos finales: w1={w1:.4f}, w2={w2:.4f}, b={b:.2f}")
    return w1, w2, b

w1, w2, b = entrenar(valores_grasa, valores_carbohidrato, valores_calorias)

```
