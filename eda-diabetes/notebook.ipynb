 # Medidas de posición no central- percentiles y cuartiles  
 ##### Mientra en la mediana se divide en 2 mitades, los cuartiles se dividen en 4 partes
 ##### Q1 (25%) → el 25% de los pacientes está por debajo de este valor
 ##### Q2 (50%) → es la mediana
 ##### Q3 (75%) → el 75% está por debajo
 ##### IQR = Q3 - Q1 → el rango donde vive el 50% central de los datos

```python
for columna in ["AGE", "BMI", "BP", "Y"]:
    q1 = db[columna].quantile(0.25)
    q2 = db[columna].quantile(0.50)
    q3 = db[columna].quantile(0.75)
    iqr = q3 - q1
    print(f"{columna} => Q1: {q1} | Q2: {q2} | Q3: {q3} | IQR: {iqr}")

# quantile(0.25) => percentil 25, Q1
# quantile(0.50) => percentil 50, equivale a la mediana
# quantile(0.75) => percentil 75, Q3
# IQR => rango intercuartílico, distancia entre Q1 y Q3
```

#Asimetría y Curtosis

#####Asimetría (skewness): si los datos están jalados hacia un lado
##### 0 = simétrico
#####> 0 = cola larga a la derecha (valores muy altos)
#####< 0 = cola larga a la izquierda (valores muy bajos)


#####Curtosis (kurtosis): qué tan puntiaguda o plana es la distribución

##### > 0 = puntiaguda, datos concentrados al centro
##### < 0 = plana, datos más dispersos

```python
for columna in ["AGE", "BMI", "BP", "Y"]:
    asimetria = db[columna].skew()
    curtosis = db[columna].kurt()
    print(f"{columna} => Asimetría: {asimetria:.4f} | Curtosis: {curtosis:.4f}")

# skew()  => mide la asimetría de la distribución
# kurt()  => mide la curtosis
# :.4f    => 4 decimales, igual que
```

#Conclusión — Asimetría y Curtosis
#####Edad (AGE)
La asimetría de -0.23 indica con una leve cola hacia la izquierda, es decir, hay pocos pacientes jóvenes que se alejan del promedio. La curtosis de -0.67 indica una distribución plana — las edades están bastante repartidas sin una concentración fuerte alrededor del promedio.
#####BMI
Es la variable más asimétrica del grupo con 0.5981, lo que indica una cola hacia la derecha — hay pacientes con BMI muy alto que se alejan del resto. La curtosis de 0.09 es casi cero, lo que significa una distribución normal en cuanto a su forma.

La asimetría de 0.29 es leve. La curtosis de -0.53 indica una distribución ligeramente plana, los valores de presión están distribuidos de forma amplia sin concentrarse demasiado en un punto.
#####Progresión de diabetes (Y)
La asimetría de 0.44 indica una cola moderada hacia la derecha — hay pacientes con progresiones muy altas que jalan la distribución. La curtosis de -0.88 es la más negativa del grupo, confirmando que es la variable más plana y dispersa de todas, consistente con lo que ya vimos en las medidas de dispersión.

#Heatmap

```python
plt.figure(figsize=(10, 8))
correlacion = db[["AGE", "BMI", "BP", "Y"]].corr()
sns.heatmap(correlacion, annot=True, fmt=".2f", cmap="coolwarm",
            linewidths=0.5, vmin=-1, vmax=1)
plt.title("Mapa de correlación entre variables")
plt.show()

# .corr()      => calcula la correlación entre todas las columnas
# annot=True   => muestra el número dentro de cada celda
# fmt=".2f"    => formato de 2 decimales para los números
# cmap="coolwarm" => azul para correlación negativa, rojo para positiva
# linewidths   => líneas entre celdas para que sea más legible
# vmin/vmax    => fija la escala de colores entre -1 y 1
```
